```python
import pandas as pd
import numpy as np
```

```python
baseline_2026 = pd.read_csv("data/control-area-balance-2026.csv", delimiter=";")
baseline_2026.head()
```

```python
# df should already have a datetime-sorted index/column and the imbalance column
df = baseline_2026.sort_values("Date Time [UTC]").reset_index(drop=True)

STEPS_PER_DAY = 96  # 15-min resolution -> 96 steps/day

df["persistence_pred"] = df["Total System Imbalance"].shift(STEPS_PER_DAY)

# drop rows where we have no prior-day value to compare against
valid = df.dropna(subset=["persistence_pred"])

mae_baseline = (valid["Total System Imbalance"] - valid["persistence_pred"]).abs().mean()
print(f"Persistence baseline MAE: {mae_baseline:.3f} MW")
```

```python
## Weekly Persistence Baseline
# df should already have a datetime-sorted index/column and the imbalance column
df = baseline_2026.sort_values("Date Time [UTC]").reset_index(drop=True)

STEPS_PER_WEEK = 96 * 7  # 15-min resolution -> 96 steps/day * 7 days/week

df["persistence_pred"] = df["Total System Imbalance"].shift(STEPS_PER_WEEK)

# drop rows where we have no prior-week value to compare against
valid = df.dropna(subset=["persistence_pred"])

mae_baseline = (valid["Total System Imbalance"] - valid["persistence_pred"]).abs().mean()
print(f"Persistence baseline MAE: {mae_baseline:.3f} MW")
```
